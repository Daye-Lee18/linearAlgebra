# 파이썬 벡터 실습 

## 벡터의 덧셈

```python
# 벡터 덧셈 
u = [1, 2, 3]
v = [4, 5, 6]

n = len(u)
w = []

for i in range(n):
    val = u[i] + v[i]
    w.append(val)
```

```python
print(w)
```

```python
def v_add(u, v):
    """
    벡터의 덧셈
    입력값: 더하고자하는 벡터 u, v
    출력값: 벡터 u, v의 덧셈 결과 w
    """
    n = len(u)
    w = []
    for idx in range(n):
        val = u[idx] + v[idx]
        w.append(val)
    return w 
```

```python
u = [1, 2, 3]
v = [4, 5, 6]
v_add(u, v)
```

## 스칼라와 벡터 곱

```python
u = [2, 4, 3]
a = 3 

n = len(u)
w = []

for i in range(n):
    val = a*u[i]
    w.append(val)
```

```python
w
```

```python
def scalar_v_mul(a,u):
    '''
    벡터의 스칼라 곱
    입력값: 스칼라 a, 벡터 리스트 u
    출력값: 스칼라 a와 벡터 리스트 u의 곱 결과 w
    '''
    n = len(u)
    w = []

    for i in range(0, n):
        val = a*u[i]
        w.append(val) 
        
    return w
```

```python
u = [2, 4, 3]
a = 3
scalar_v_mul(a, u)
```

## 벡터의 곱셈 

```python
def v_mul(u, v):
    """
    벡터의 원소 곱
    입력값: 원소 곱 하고자할 벡터 리스트 u,v 
    출력값: 벡터 u, v의 원소 곱 결과 w 
    """
    n = len(u)
    w = []
    for idx in range(n):
        val = u[idx] * v[idx]
        w.append(val)
    return w
```

```python
u = [1,2,4]
v = [7,3,2]
v_mul(u,v)
```

## 벡터의 나눗셈

```python
def v_div(u, v):
    """
    벡터의 원소 나눗셈
    입력값: 원소 나눗셈 하고자할 벡터 리스트 u, v
    출력값: 벡터 u, v의 원소 나눗셈 결과 w 
    """
    n = len(u)
    w = []

    for idx in range(n):
        val = u[idx] / v[idx]
        w.append(val)
    return w
```

```python
u = [6,5,9]
v = [2,2,-3]
v_div(u,v)
```
