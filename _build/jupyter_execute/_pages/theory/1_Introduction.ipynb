# 선형대수학의 개요 

## 기본적인 수학 개념 

### 집합 

- 집합 (Set): 특정한 조건을 만족하는 어떤 대상들의 모임 
- 원소 (Element): 집합에 포함되는 각 대상 
- 집합의 표현 
  - 원소 나열법 (tabular form): 중괄호 { } 안에 원소 나열
  - 조건 제시법 (set builder  form): 집합에 속하는 원소들의 공통 성질을 조건으로 제시 
- 원소 x가 집합 X에 포함된다. 
  - $x \in X$
- 부분 집합 (subset): 집합 A의 모든 원소가 집합 B에 속할 때 A는 B의 부분집합 
  - $A \subset B$
- 공집합 (empty): 어떠한 원소도 포함되지 않는 집합 
  - $\empty$, {}
- 집합의 연산 
  - 합집합(union) $ \cup $
  - 교집합(intersection) $ \cap $
  - 차집합 (difference) $ -$

### 사상 

- 사상 (mapping): 집합 A와 B에 대해서 A의 각 원소가 B의 어떤 원소 하나에 대응될 때, 이 관계를 A에서 B로서의 사상 (mapping) f 라고 한다. 
  - $f: A \rightarrow B$
  - 정의역 /정의구역 (domain):
  - 공역 / 공번역 (codomain): 
  - 상 (image): $ f(a_{i}) = b_{i}$
  - 치역 (range): 정의역 원소의 상을 모아 놓은 집합 
  - 전사 (surjection), 위로의 사상 (onto mapping): 공역과 치역이 동일한 사상 
  - 단사 (injection), 일대일 사상 (one-to-one mapping)
  - 전단사 (bijection), 일대일 대응 (one-to-one correscpondence)
  - 역사상 (inverse mapping)
